# Session 03: Data cleaning & wrangling

Practice companion for New. **Synthetic demonstration data**, not the lecturer's original dataset. Run the code cells top to bottom. The original slide snippets remain available separately in the website.

Requires numpy, pandas, matplotlib, seaborn, scikit-learn, and plotly. Install missing packages with `%pip install numpy pandas matplotlib seaborn scikit-learn plotly`.

```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(42)
n = 120
df = pd.DataFrame({
    'student_id': np.arange(1, n+1),
    'name': [f'Student {i}' for i in range(1, n+1)],
    'faculty': rng.choice(['Engineering', 'Science', 'Arts'], n),
    'age': rng.integers(18, 25, n),
    'study_hours': rng.uniform(1, 10, n).round(1),
    'gpa': np.clip(rng.normal(3.0, 0.5, n), 0, 4).round(2),
    'passed_exam': rng.random(n) > 0.2,
})
raw = pd.concat([df, df.iloc[:3]], ignore_index=True)
raw.loc[::9, 'age'] = np.nan
raw['gpa'] = raw['gpa'].astype(str)
raw['faculty'] = ' ' + raw['faculty'].str.lower() + ' '
def clean(raw):
    result = raw.copy().drop_duplicates('student_id')
    result['gpa'] = pd.to_numeric(result['gpa'], errors='coerce')
    result['age'] = result['age'].fillna(result['age'].median())
    result['faculty'] = result['faculty'].str.strip().str.title()
    return result
clean_df = clean(raw)
print('Rows before / after:', len(raw), len(clean_df))
print(clean_df.isna().sum())
assert clean_df.student_id.is_unique
```

## Watch out
A median is not automatically the right replacement. For predictive models, learn imputation values on the training data only. Use merge validation when you expect a many-to-one or one-to-one relationship.
